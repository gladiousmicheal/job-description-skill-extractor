```python
job_description = """
We are looking for a Data Analyst skilled in Python, SQL, Excel,
Power BI, and Machine Learning. Experience in data visualization
and data cleaning is preferred.
"""

print(job_description)
```

```python
def extract_skills(job_description):
    skills_list = [
        "Python",
        "SQL",
        "Excel",
        "Power BI",
        "Machine Learning",
        "Data Visualization",
        "Data Cleaning"
    ]

    found_skills = []

    for skill in skills_list:
        if skill.lower() in job_description.lower():
            found_skills.append(skill)

    return found_skills
```

```python
extract_skills("""
We need a candidate with Python, SQL, and Power BI experience.
""")
```

```python
job1 = "Looking for a Python developer with SQL skills."
job2 = "Must know Excel and Power BI."
job3 = "Looking for a Java developer."

print("Job 1:", extract_skills(job1))
print("Job 2:", extract_skills(job2))
print("Job 3:", extract_skills(job3))
```
